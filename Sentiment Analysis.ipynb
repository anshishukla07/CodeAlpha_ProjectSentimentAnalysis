```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# 1. Fake Amazon reviews
amazon_data = {
    'Review': [
        "This phone is amazing! Best camera and long battery life.",
        "Completely waste of money. The screen broke on day one.",
        "It's an okay product. Delivery was fast but features are average.",
        "Superb performance, highly recommended to everyone!",
        "Very bad customer support. The device keeps freezing."
    ]
}
df = pd.DataFrame(amazon_data)

# 2. Sentiment engine 
sia = SentimentIntensityAnalyzer()

# 3. Function automatically Positive, Negative,Neutral 
def check_sentiment(text):
    score = sia.polarity_scores(text)
    if score['compound'] >= 0.05:
        return 'Positive'
    elif score['compound'] <= -0.05:
        return 'Negative'
    else:
        return 'Neutral'

# 4. Pure data function apply
df['Result'] = df['Review'].apply(check_sentiment)

# 5. Countplot visual 
plt.figure(figsize=(6, 4))
sns.countplot(x='Result', data=df, palette=['green', 'red', 'gray'])
plt.title("Amazon Reviews - Sentiment Analysis Report", fontsize=12, fontweight='bold')
plt.xlabel("Sentiment Category")
plt.ylabel("Number of Reviews")
plt.show()
```
